# Plugin Packaging & Entry Points
This lesson demonstrates how GeneCoder discovers plugins via Python entry points.


1. Install the example codec from `plugins-examples/example_codec`.
2. Examine its `pyproject.toml` to see the `genecoder.plugins` entry.
3. Use `load_plugins()` to register entry points and inspect the codec registry.


```
from genecoder.plugins import load_plugins
load_plugins()
from genecoder import CODEC_REGISTRY
list(CODEC_REGISTRY)
```
